```python
import torch
import torch.nn as nn
```

```python
class RNNModel(nn.Module):
    def __init__(self):
        super(RNNModel,self).__init__()
        self.run = nn.RNN(input_size=5,hidden_size=3,batch_first=True)
        self.fc = nn.Linear(3,1)
        self.sigmoid = nn.Sigmoid()
    
    def forward(self,x):
        out, _ = self.run(x)
        out = out[:,-1,:]
        out = self.fc(out)
        out = self.sigmoid(out)
        return out    
        
```
