```python
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd
```

```python

#& Documents (corpus)
corpus = [
    "Student watch youtube youtube youtube"
    ,"youtube watch youtube",
    "student write comments",
    "youtube write comments"
]
```

```python

# create bag of words model
cv = CountVectorizer(binary=True)

# learn the vocabulary dictionary and return term-document matrix
X = cv.fit_transform(corpus)
```

```python

# show vocabulary
cv.get_feature_names_out()
```

```python

# show bag of words matrix
X.toarray()
```

```python

bow_df = pd.DataFrame(X.toarray(), columns=cv.get_feature_names_out(),index=["D1","D2","D3","D4"])

bow_df 
```

## TF-IDF (Term Frequency - Inverse Document Frequency)

Unlike the binary Bag-of-Words model above (which only records presence/absence of a word), **TF-IDF** weighs each term by how important it is to a document relative to the whole corpus:

- **TF (Term Frequency)**: how often a word appears in a document.
- **IDF (Inverse Document Frequency)**: down-weights words that appear in many documents (e.g. "youtube" appears everywhere) and up-weights rarer, more distinctive words.

We reuse the same `corpus` defined earlier so we can directly compare the Bag-of-Words matrix with the TF-IDF matrix.

```python

from sklearn.feature_extraction.text import TfidfVectorizer

# create the TF-IDF model using the SAME corpus as the Bag-of-Words example
tfidf = TfidfVectorizer()

# learn the vocabulary and idf, then return the term-document TF-IDF matrix
X_tfidf = tfidf.fit_transform(corpus)

# show vocabulary (should match the CountVectorizer vocabulary)
tfidf.get_feature_names_out()
```

```python

# show TF-IDF matrix as a readable DataFrame, using the same document labels (D1..D4)
tfidf_df = pd.DataFrame(
    X_tfidf.toarray(),
    columns=tfidf.get_feature_names_out(),
    index=["D1", "D2", "D3", "D4"]
)

tfidf_df.round(3)
```

### Comparing Bag-of-Words vs TF-IDF

Notice how in `bow_df`, the word **"youtube"** (which appears in 3 of the 4 documents) has the same weight as rarer words like **"comments"**. In `tfidf_df`, "youtube" gets a *lower* weight because it is common across the corpus, while distinctive words like "comments" or "write" get a relatively *higher* weight in the documents where they appear. This is the key benefit of TF-IDF over plain Bag-of-Words / CountVectorizer.
