```python
import re
```

```python
text = "The rain in Spain. I have 1 apple and 2 oranges."

# Search for the word "rain" in the text
match = re.search(r"rain", text)
print("Found:", match.group())

# search the . character in the text
match_dot = re.search(r"\.", text)
print("Found dot:", match_dot.group())

# Search for all digits in the text
matches_digits = re.findall(r"\d", text)
print("Found digits:", matches_digits)
            
```
