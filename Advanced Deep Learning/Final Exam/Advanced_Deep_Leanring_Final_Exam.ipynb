## Q1: One-Hot Encoding

```python
from sklearn.preprocessing import OneHotEncoder

words = [["cat"], ["dog"], ["bird"], ["cat"]]

encoder = OneHotEncoder()

encoded = encoder.fit_transform(words)

print(encoded.toarray())
```

## Q2: TF-IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer

documents = [
    "I love NLP",
    "NLP is amazing",
    "Machine learning is fun"
]

vectorizer = TfidfVectorizer()

tfidf_matrix = vectorizer.fit_transform(documents)

print(tfidf_matrix.toarray())
```

## Q11: Stopword Removal

```python
from nltk.corpus import stopwords
import nltk

nltk.download('stopwords')

words = ["I", "love", "to", "learn", "NLP"]

stop_words = set(stopwords.words("english"))

filtered = [word for word in words if word.lower() not in stop_words]

print(filtered)
```

## Q12: Stemming

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

words = ["playing", "played", "plays"]

stems = [stemmer.stem(word) for word in words]

print(stems)
```

## Q13: Lemmatization

```python
from nltk.stem import WordNetLemmatizer
import nltk

nltk.download('wordnet')

lemmatizer = WordNetLemmatizer()

words = ["running", "better", "cars"]

lemmas = [lemmatizer.lemmatize(word) for word in words]

print(lemmas)
```

## Q20: Complete NLP Preprocessing

```python
import re

text = "Hello!!! Visit https://abc.com Email: student@gmail.com 123"

text = text.strip()

text = re.sub(r"https?://\S+", "", text)
text = re.sub(r"\S+@\S+", "", text)
text = re.sub(r"\d+", "", text)
text = re.sub(r"[^\w\s]", "", text)
text = re.sub(r"\s+", " ", text).strip()

print(text)
```

Explain the difference between **CBOW (Continuous Bag of Words)** and **Skip-Gram** architectures.

Your answer should include:

1. **How each model predicts words:**
   - **CBOW** predicts a target word from its surrounding context words. It takes multiple context words as input and tries to predict the word that fits in the middle.
   - **Skip-Gram** does the opposite: it takes a single target word as input and tries to predict the surrounding context words. Each word is used to predict its neighbors.

2. **Which model performs better on small datasets:**
   - **CBOW** performs better on small datasets because it averages context words, which smooths out noise and requires less training data to learn meaningful representations.

3. **Which model is better for learning rare words:**
   - **Skip-Gram** is better for rare words because it gives each occurrence of a word multiple chances to be learned (once for each context pair), whereas CBOW averages the context and can dilute the signal from rare words.

4. **One practical advantage of Word2Vec over One-Hot Encoding:**
   - Word2Vec produces dense, low-dimensional vectors that capture semantic relationships between words (e.g., `king - man + woman ≈ queen`), while One-Hot Encoding creates sparse, high-dimensional vectors with no notion of similarity and cannot capture meaning.
