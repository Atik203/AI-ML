# Question 01: [ Marks 10 ]
Explain gradient descent in detail. What is the main idea behind it? How does it help in minimizing the loss function? Explain the role of learning rate and describe what happens if the learning rate is too high or too low.

**Write Answer 01:**

> Gradient Descent is an optimization algorithm used to minimize the loss function of a model.
> Main idea: move parameters in the opposite direction of the gradient, because gradient points to the steepest increase.

If model parameters are $\theta$ and loss is $J(\theta)$, update rule is:

$$\theta_{t+1}=\theta_t-\eta\nabla J(\theta_t)$$

Here, $\eta$ is the learning rate.

How it minimizes loss:
- At each step, gradient tells us how loss changes.
- Subtracting the gradient moves parameters toward lower loss.
- Repeating this many times brings us near a minimum (global or local).

Role of learning rate ($\eta$):
- Too high: updates jump too far, loss may oscillate or diverge.
- Too low: training becomes very slow and may get stuck before good convergence.
- Good learning rate: stable and reasonably fast decrease in loss.

# Question 02: [ Marks 10 ]

What is forward propagation in an artificial neural network? Explain step by step how data moves from the input layer to the output layer.

## Write Answer 02:

> Forward propagation means passing input data through the network (layer by layer) to produce prediction.

Step-by-step flow:
1. Input layer receives features, e.g., $x=[x_1,x_2,\dots,x_n]$.
2. First hidden layer computes weighted sum: $z^{[1]}=W^{[1]}x+b^{[1]}$.
3. Activation is applied: $a^{[1]}=g(z^{[1]})$.
4. Next hidden layers repeat same process:
   $z^{[l]}=W^{[l]}a^{[l-1]}+b^{[l]}$, then $a^{[l]}=g(z^{[l]})$.
5. Output layer gives final prediction $\hat{y}$ (for example probability/class score).
6. During training, this $\hat{y}$ is compared with true label $y$ to compute loss.

So, forward propagation is basically:
**Input -> Linear transform -> Activation -> ... -> Output prediction**.

# Question 03: [ Marks 15 ]
Why do we need activation functions in neural networks? What would happen if we only used linear transformations without any activation function?

## Write Answer 03:

> Activation functions are needed to introduce non-linearity.

Without activation functions:
- Every layer does only linear transformation.
- Stacking many linear layers still becomes one single linear function.
- So deep network would behave like simple linear model:

$$y = Wx + b$$

That means it cannot learn complex patterns like curves, XOR-like structure, image/text complexity, etc.

With activation functions (ReLU, sigmoid, tanh), network can learn non-linear decision boundaries and represent much richer functions.

# Question 04: [ Marks 15 ]

Explain the sigmoid activation function. Where is it used? What are its advantages and disadvantages? What problem does it cause during training?

## Write Answer 04:

> Sigmoid activation converts any real value to range $(0,1)$.

$$\sigma(z)=\frac{1}{1+e^{-z}}$$

Where it is used:
- Commonly in output layer of binary classification to represent probability.

Advantages:
- Output is bounded between 0 and 1 (easy probabilistic interpretation).
- Smooth and differentiable.

Disadvantages:
- For very large positive/negative $z$, gradient becomes very small.
- Not zero-centered output (can slow optimization).
- Usually slower convergence than ReLU in deep hidden layers.

Training problem caused:
- **Vanishing gradient problem**. In deep networks, gradients shrink across layers, so earlier layers learn very slowly.

# Question 05: [ Marks 10 ]
Explain the softmax activation function in simple words. Why is it used in multi-class classification problems? How is it different from sigmoid?

## Write Answer 05:

> Softmax turns raw scores (logits) into class probabilities for multiple classes.

For class $i$:

$$\text{softmax}(z_i)=\frac{e^{z_i}}{\sum_{j=1}^{K} e^{z_j}}$$

Why used in multi-class classification:
- It gives probability for each class.
- All probabilities sum to 1.
- Model can choose class with highest probability.

Difference from sigmoid:
- Sigmoid works independently per output (good for binary or multi-label).
- Softmax links all class outputs together (good for mutually exclusive classes).
- In short: sigmoid = independent probabilities, softmax = normalized competition among classes.

# Question 06: [ Marks 20 ]

Explain the structure of an Artificial Neural Network in detail. Describe the role of the input layer, hidden layers, and output layer. How do weights, biases, and activation functions work together to transform input data into predictions? Illustrate your explanation with a simple example.

## Write Answer 06:

> An Artificial Neural Network (ANN) is a layered model that transforms input features into predictions through weighted computations and non-linear activations.

Structure and role of layers:
1. Input layer:
- Takes raw features (e.g., age, income, pixel values).
- No heavy computation, mainly data entry point.

2. Hidden layer(s):
- Perform transformation and feature extraction.
- Each neuron computes:

$$z = Wx + b,\quad a = g(z)$$

where $W$ = weights, $b$ = bias, $g$ = activation function.

3. Output layer:
- Produces final prediction.
- Activation depends on task: sigmoid (binary), softmax (multi-class), linear (regression).

How weights, biases, and activation work together:
- Weights control influence of each input.
- Bias shifts the activation threshold.
- Activation adds non-linearity so network learns complex patterns.

Simple example:
- Suppose input $x=[x_1,x_2]$.
- Hidden neuron: $z_1=w_{11}x_1+w_{12}x_2+b_1$, then $a_1=\text{ReLU}(z_1)$.
- Another hidden neuron gives $a_2$.
- Output neuron: $\hat{y}=\sigma(v_1a_1+v_2a_2+b_o)$.
- If $\hat{y}\approx 1$, model predicts class 1; if near 0, class 0.

So ANN prediction is a chain of transformations from simple inputs to meaningful decision.
