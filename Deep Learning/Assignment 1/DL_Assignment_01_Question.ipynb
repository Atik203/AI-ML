# Question 01: [ Marks 10 ]

What is a perceptron?
Explain its three main components and their roles.

## Write Answer 01:

Perceptron is the simplest neural network model.
It is like one artificial neuron.
It is used for basic classification, mostly 0 or 1.

A perceptron takes input values, makes a weighted sum, and then gives output.
The output depends on a threshold (step) function.

Main 3 parts:
- Inputs (`x1, x2, ...`): data values or features.
- Weights and bias (`w`, `b`): weights show importance of each input, bias shifts the decision line.
- Activation (step): converts the sum into final class output.

# Question 02: [ Marks 10 ]

What is a decision boundary in a perceptron?
Explain how the equation
w₁x₁ + w₂x₂ + b = 0
represents a decision boundary geometrically.

## Write Answer 02:

Decision boundary is the line that separates classes.
In perceptron, this boundary is created by a linear equation.

For 2 inputs:
`w1x1 + w2x2 + b = 0`

Meaning:
- `w1x1 + w2x2 + b > 0`: one class (for example class 1)
- `w1x1 + w2x2 + b < 0`: other class (for example class 0)
- `w1x1 + w2x2 + b = 0`: exactly on the boundary line

So geometrically this equation is the separating straight line in 2D space.

# Question 03: [ Marks 15 ]

Why is the perceptron called a linear classifier?
Explain your answer using the concept of linear separability and logical gate examples such as AND, OR, and XOR.

## Write Answer 03:

Perceptron is called a linear classifier.
Because it uses one linear equation:
`w^T x + b = 0`.
In 2D, this is a straight line.

If data can be separated by one straight line, perceptron can learn it.
This is called linear separability.

Examples:
- AND: linearly separable, so perceptron works.
- OR: linearly separable, so perceptron works.
- XOR: not linearly separable, so perceptron does not work.

So perceptron is good for simple linear problems only.

# Question 04: [ Marks 15 ]

Write the perceptron weight update rule and explain it intuitively.
Why are the weights updated only when the prediction is incorrect?

## Write Answer 04:

Perceptron update rule:

`w_new = w_old + eta * (y - y_hat) * x`

`b_new = b_old + eta * (y - y_hat)`

Here:
- `eta`: learning rate
- `y`: true label
- `y_hat`: predicted label

Intuition:
- If prediction is correct, error `(y - y_hat)` is 0, so no update.
- If prediction is wrong, weights and bias are updated.
- This update moves the decision boundary toward correct classification.

Weights are updated only when prediction is incorrect because the model learns from mistakes.

# Question 05: [ Marks 10 ]

Why does a single-layer perceptron fail to solve the XOR problem?
What does this limitation tell us about the need for multilayer neural networks?

## Write Answer 05:

Single-layer perceptron fails on XOR.
Reason: XOR points cannot be separated by one straight line.
A single-layer perceptron can make only one linear boundary.

So one layer is not enough for XOR type problems.
This tells us we need multilayer neural networks (MLP).
With hidden layers and nonlinear activation, MLP can create nonlinear boundaries and solve XOR.

# Question 06: [ Marks 20 ]

What is meant by the “perceptron Learning Rule” or weight adjustment method?
Why can adjusting one weight affect the classification of other points?

## Write Answer 06:

Perceptron Learning Rule means we adjust weights and bias when model prediction is wrong.
The goal is to reduce error step by step.

Rule:

`w = w + eta * (y - y_hat) * x`

`b = b + eta * (y - y_hat)`

Why one weight affects other points:
All data points are classified by the same equation `w^T x + b`.
If one weight changes, the boundary line also changes.
So many points can move from one side of the boundary to the other side.
That is why changing one weight can affect classification of other points.

## Write Answer 07:

Main limitations of single-layer perceptron:
- Works only for linearly separable data
- Cannot solve XOR and similar nonlinear problems
