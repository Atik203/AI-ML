# Question 01: [ Marks 10 ]

What is a perceptron?
Explain its three main components and their roles.

## Write Answer 01:

Perceptron is the simplest neural network model.
It is like one artificial neuron.
It does classification (0 or 1).

Main 3 parts:
- Inputs (`x1, x2, ...`): data values.
- Weights and bias (`w`, `b`): control importance and shift boundary.
- Activation (step): gives final output class.

# Question 02: [ Marks 10 ]

What is a decision boundary in a perceptron?
Explain how the equation
w₁x₁ + w₂x₂ + b = 0
represents a decision boundary geometrically.

## Write Answer 02:

Decision boundary is the line that separates classes.

For 2 inputs:
`w1x1 + w2x2 + b = 0`

Meaning:
- `> 0`: one class (ex: 1)
- `< 0`: other class (ex: 0)
- `= 0`: on the boundary line

So this equation is the separating line.

# Question 03: [ Marks 15 ]

Why is the perceptron called a linear classifier?
Explain your answer using the concept of linear separability and logical gate examples such as AND, OR, and XOR.

## Write Answer 03:

Perceptron is a linear classifier.
Because it uses one linear equation:
`w^T x + b = 0`.
In 2D, this is a straight line.

Examples:
- AND: works
- OR: works
- XOR: does not work

Why XOR fails:
XOR is not linearly separable.

# Question 04: [ Marks 15 ]

Write the perceptron weight update rule and explain it intuitively.
Why are the weights updated only when the prediction is incorrect?

## Write Answer 04:

Perceptron update rule:
`w_new = w_old + eta * (y - y_hat) * x`
`b_new = b_old + eta * (y - y_hat)`

Here:
- `eta`: learning rate
- `y`: true label
- `y_hat`: predicted label

If prediction is correct, error is 0, so no update.
If prediction is wrong, weights change.
This moves boundary in correct direction.

## Write Answer 05:

Single-layer perceptron fails on XOR.
Because XOR cannot be separated by one straight line.

This shows one layer is not enough for some problems.
So we use multilayer neural network (MLP).
MLP can make nonlinear boundaries and solve XOR.

# Question 06: [ Marks 20 ]

What is meant by the “perceptron Learning Rule” or weight adjustment method?
Why can adjusting one weight affect the classification of other points?

## Write Answer 06:

Perceptron learning rule means:
change weights when prediction is wrong.

Rule:
`w = w + eta * (y - y_hat) * x`
`b = b + eta * (y - y_hat)`

Why one weight affects other points:
All points use the same boundary equation.
If one weight changes, boundary line changes.
So many points can change class.
