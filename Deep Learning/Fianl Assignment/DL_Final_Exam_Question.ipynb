# Question 01: [ Marks 10 ]

## Explain the concept of backpropagation in detail. Describe how the error is propagated from the output layer to the hidden layers. Derive the mathematical expressions for weight updates using the chain rule, and explain how learning rate affects the training process.


# Write Answer 01:

Backpropagation is the algorithm used to train neural networks by computing gradients of the loss with respect to all weights and biases.

It works in two phases:
1. **Forward pass**: compute prediction and loss.
2. **Backward pass**: propagate error from output layer to hidden layers using chain rule.

Assume layer `l`:
$$
z^{[l]}=W^{[l]}a^{[l-1]}+b^{[l]},\quad a^{[l]}=g^{[l]}(z^{[l]})
$$

For output layer `L`, error term:
$$
\delta^{[L]}=\frac{\partial L}{\partial z^{[L]}}
$$

For hidden layers (`l = L-1 ... 1`):
$$
\delta^{[l]}=((W^{[l+1]})^T\delta^{[l+1]})\odot g'^{[l]}(z^{[l]})
$$
This equation shows how error is propagated backward from output to hidden layers.

Now gradients are:
$$
\frac{\partial L}{\partial W^{[l]}}=\delta^{[l]}(a^{[l-1]})^T,
\quad
\frac{\partial L}{\partial b^{[l]}}=\delta^{[l]}
$$

Gradient-descent update:
$$
W^{[l]} \leftarrow W^{[l]}-\eta\frac{\partial L}{\partial W^{[l]}},
\quad
b^{[l]} \leftarrow b^{[l]}-\eta\frac{\partial L}{\partial b^{[l]}}
$$

Where $\eta$ is learning rate.

### Effect of learning rate
- **Too high:** overshooting, oscillation, or divergence.
- **Too low:** very slow learning.
- **Proper value:** stable and fast convergence.

So backpropagation efficiently computes gradients layer-by-layer and updates parameters to minimize the loss.

# Question 02: [ Marks 10 ]

## Consider a neural network with one hidden layer using sigmoid activation. Derive the complete backpropagation equations for updating weights and biases. Show all steps clearly, including forward pass, loss calculation, and backward pass.

## Write Answer 02:

Consider one hidden-layer network (sigmoid hidden and sigmoid output):
$$
a^{[1]}=\sigma(z^{[1]}),\quad z^{[1]}=W^{[1]}x+b^{[1]}
$$
$$
\hat{y}=a^{[2]}=\sigma(z^{[2]}),\quad z^{[2]}=W^{[2]}a^{[1]}+b^{[2]}
$$

Use binary cross-entropy loss:
$$
L = -\left(y\log \hat{y} + (1-y)\log(1-\hat{y})\right)
$$

### Forward pass
1. Compute hidden pre-activation: `z[1] = W[1]x + b[1]`
2. Hidden activation: `a[1] = sigmoid(z[1])`
3. Output pre-activation: `z[2] = W[2]a[1] + b[2]`
4. Prediction: `a[2] = sigmoid(z[2])`
5. Compute loss `L`.

### Backward pass
For sigmoid + BCE:
$$
\delta^{[2]} = \frac{\partial L}{\partial z^{[2]}} = a^{[2]} - y
$$

Output-layer gradients:
$$
\frac{\partial L}{\partial W^{[2]}} = \delta^{[2]}(a^{[1]})^T,
\quad
\frac{\partial L}{\partial b^{[2]}} = \delta^{[2]}
$$

Hidden-layer error:
$$
\delta^{[1]} = (W^{[2]})^T\delta^{[2]} \odot \sigma'(z^{[1]})
$$
where
$$
\sigma'(z)=\sigma(z)(1-\sigma(z))
$$

Hidden-layer gradients:
$$
\frac{\partial L}{\partial W^{[1]}} = \delta^{[1]}x^T,
\quad
\frac{\partial L}{\partial b^{[1]}} = \delta^{[1]}
$$

### Parameter update
$$
W^{[l]} \leftarrow W^{[l]} - \eta \frac{\partial L}{\partial W^{[l]}},
\quad
b^{[l]} \leftarrow b^{[l]} - \eta \frac{\partial L}{\partial b^{[l]}}
$$

These are the complete backprop equations for a one-hidden-layer sigmoid network.

# Question 03: [Marks 5]

## Discuss the vanishing gradient problems in deep neural networks. Explain why they occur during backpropagation, how activation functions contribute to this issue, and suggest at least three techniques to overcome these problems.


## Write Answer 03:

Vanishing gradient means gradients become extremely small as they are propagated backward through many layers.

### Why it happens
During backpropagation, gradients are multiplied repeatedly by derivatives of activation functions and weights:
$$
\frac{\partial L}{\partial W^{[1]}} \propto \prod_{l=2}^{L} \left(W^{[l]}\cdot g'(z^{[l-1]})\right)
$$
If many terms are $< 1$, the product shrinks toward $0$.

### Role of activation functions
- **Sigmoid / Tanh:** in saturated regions derivatives are very small.
  - Sigmoid derivative max is $0.25$.
- Repeated multiplication of such small values across deep layers leads to almost zero gradient in early layers.

### Effects
- Early layers learn very slowly.
- Training becomes unstable/slow.
- Deep networks may fail to converge properly.

### Techniques to overcome vanishing gradient (at least 3)
1. Use **ReLU-family** activations (ReLU, LeakyReLU, ELU).
2. Use proper **weight initialization** (He/Xavier).
3. Apply **Batch Normalization**.
4. Use **Residual connections** (ResNet skip connections).
5. Use adaptive optimizers and learning-rate scheduling.

These methods improve gradient flow and make deep networks train effectively.

# Question 04: [ Marks 5 ]

## Describe the architecture of an Artificial Neural Network (ANN) in detail. Explain the role of each component (neurons, weights, bias, activation functions). Also discuss how increasing the number of hidden layers and neurons affects model performance and overfitting.


## Write Answer 04:

An Artificial Neural Network (ANN) is composed of layers of neurons that transform input data step by step.

### Main components
1. **Neurons:** computation units that receive input, apply weighted sum and activation.
2. **Weights (`W`):** control importance of each input feature.
3. **Bias (`b`):** shifts the neuron activation threshold.
4. **Activation function (`g`)**: introduces non-linearity.

Each neuron computes:
$$
z = Wx + b, \quad a = g(z)
$$

### Layer-wise structure
- **Input layer:** receives raw features.
- **Hidden layer(s):** extract intermediate patterns.
- **Output layer:** gives final prediction (class/probability/value).

### Effect of increasing hidden layers/neurons
- **More layers/neurons (positive):** higher capacity, can model complex patterns.
- **Too many (negative):** may overfit training data, needs regularization (dropout, L2, early stopping).
- **Trade-off:** choose architecture based on dataset size and complexity.

So ANN performance improves with sufficient capacity, but excessive complexity without proper control can reduce generalization.

# Question 05: [ Marks 10 ]

## Compare Artificial Neural Networks (ANN) and Convolutional Neural Networks (CNN) in detail. Discuss their architectures, working principles, advantages, limitations, and typical use cases such as image processing, tabular data, and pattern recognition.

## Write Answer 05:

ANN and CNN are both neural-network models, but they are designed for different types of data and structure.

### Architecture difference
- **ANN (Fully Connected):** each neuron connects to all neurons in previous layer.
- **CNN:** uses convolution layers, pooling layers, and then fully connected layers.

### Working principle
- **ANN:** learns global relationships directly from flattened feature vectors.
- **CNN:** learns spatial/local patterns using filters (edges $\rightarrow$ texture $\rightarrow$ object parts).

### Advantages
- **ANN:** simple, good for tabular data and smaller feature spaces.
- **CNN:** excellent for images due to local receptive fields and parameter sharing.

### Limitations
- **ANN:** too many parameters for image input, ignores spatial structure.
- **CNN:** requires more design choices (kernel size, stride, channels), more compute than small ANN.

### Typical use cases
- **Image processing / pattern recognition:** CNN (classification, detection, medical image analysis).
- **Tabular data:** ANN (credit scoring, churn prediction, forecasting with engineered features).
- **General low-dimensional numeric input:** ANN is often sufficient.

So, CNN is a specialized neural architecture for grid-like data (images), while ANN is a general-purpose baseline for vector/tabular data.

# Question 06: [ Marks 10 ]

## Explain different activation functions used in neural networks, including Sigmoid, Tanh, and ReLU. Provide their mathematical formulas, advantages, disadvantages, and suitable use cases. Also discuss why ReLU is widely used in deep learning.

## Write Answer 06:

Activation functions add non-linearity so neural networks can learn complex patterns.

### 1) Sigmoid
Formula:
$$
\sigma(x)=\frac{1}{1+e^{-x}}
$$
- Output range: `(0,1)`
- **Advantages:** smooth, probabilistic interpretation.
- **Disadvantages:** saturates at large $|x|$ $\rightarrow$ very small gradient, not zero-centered.
- **Use cases:** output layer for binary classification.

### 2) Tanh
Formula:
$$
\tanh(x)=\frac{e^x-e^{-x}}{e^x+e^{-x}}
$$
- Output range: $(-1,1)$
- **Advantages:** zero-centered, usually better than sigmoid in hidden layers.
- **Disadvantages:** still suffers saturation/vanishing gradient for large $|x|$.
- **Use cases:** some hidden layers in shallow/medium networks.

### 3) ReLU
Formula:
$$
\mathrm{ReLU}(x)=\max(0,x)
$$
- Output range: $[0, +\infty)$
- **Advantages:** simple, fast, sparse activation, mitigates vanishing gradient on positive side.
- **Disadvantages:** dead ReLU problem (neurons can get stuck at 0 for negative inputs).
- **Use cases:** most hidden layers in deep CNN/ANN models.

### Why ReLU is widely used
- Faster training due to simple computation.
- Better gradient flow in deep networks compared to sigmoid/tanh.
- Works very well in practice for image, text, and many deep-learning tasks.

## Describe the convolution operation in CNNs in detail. Explain how filters (kernels) work, how feature maps are generated, and how parameters like stride and padding affect the output. Provide a step-by-step example using a small input matrix.
Note :  You've to attach an image here.



## Write Answer 07:

Convolution is the core operation in CNN where a small filter (kernel) slides over an input image and computes dot products.

If input is `X` and kernel is `K`, then each output pixel is:
$$
Y(i,j)=\sum_m\sum_n X(i+m, j+n)K(m,n)
$$

### How filters work
- A kernel (e.g., $3\times3$) has learnable weights.
- The same kernel is applied at all spatial locations (parameter sharing).
- Different kernels learn different patterns (edges, texture, corners).

### Feature map generation
1. Place kernel on the first local patch.
2. Multiply corresponding values and sum.
3. Store result in output position.
4. Move kernel according to stride and repeat.

### Effect of stride and padding
- **Stride ($S$):** step size of kernel movement.
  - Larger stride $\rightarrow$ smaller output.
- **Padding ($P$):** zeros added around border.
  - $P=0$ (valid) shrinks output.
  - $P=1$ with $3\times3$ can keep size same (if $S=1$).

Output size formula:
$$
\text{Out} = \left\lfloor \frac{N + 2P - F}{S} \right\rfloor + 1
$$

### Small step-by-step example
Input (4x4):
$$
\begin{bmatrix}
1&2&0&1\\
3&1&2&2\\
0&1&3&1\\
2&2&1&0
\end{bmatrix}
$$
Kernel (2x2):
$$
\begin{bmatrix}
1&0\\
0&1
\end{bmatrix}
$$
With stride=1, padding=0:
- Top-left output = `1*1 + 2*0 + 3*0 + 1*1 = 2`
- Next positions are computed similarly.

So convolution transforms raw pixels into feature maps that represent useful visual patterns.

# Question 08: [ Marks 10 ]

## Explain the concept of pooling in CNNs. Compare MaxPooling and AveragePooling in terms of operation, advantages, and impact on feature extraction. Discuss how pooling helps in reducing overfitting and computational complexity.

## Write Answer 08:

Pooling is a downsampling operation used in CNNs to reduce spatial size ($H \times W$) of feature maps while keeping important information.

### MaxPooling vs AveragePooling
- **MaxPooling:** takes maximum value from each window (e.g., $2\times2$).  
  It keeps the strongest/most activated feature.
- **AveragePooling:** takes average value from each window.  
  It keeps general/background information.

### Comparison
- **Operation:** max value vs mean value.
- **Feature extraction impact:**
  - MaxPooling highlights prominent edges/textures.
  - AveragePooling gives smoother representation.
- **Noise behavior:**
  - MaxPooling can be robust to small local noise.
  - AveragePooling may blur sharp features.

### Why pooling is useful
1. **Reduces computation:** smaller feature maps $\rightarrow$ fewer operations in next layers.
2. **Helps reduce overfitting:** fewer activations and parameters in later layers.
3. **Provides translation tolerance:** small shifts in object position affect output less.

In many vision tasks, **MaxPooling** is more common because it preserves the most informative activations.

# Question 09: [ Marks 10 ]

## Given an input image of size 128×128×3, explain step-by-step how the dimensions change after multiple convolution and pooling layers. Clearly show calculations for output size after each layer using appropriate formulas.

# Write Answer 09:

For convolution output size (per spatial dimension):

$$
\text{Out} = \left\lfloor \frac{N + 2P - F}{S} \right\rfloor + 1
$$

where $N$ = input size, $F$ = filter size, $P$ = padding, $S$ = stride.

Given input image: **$128 \times 128 \times 3$**

Assume this sequence:
1. Conv1: $3\times3$, stride $=1$, padding $=1$, filters $=32$  
2. MaxPool1: $2\times2$, stride $=2$  
3. Conv2: $3\times3$, stride $=1$, padding $=1$, filters $=64$  
4. MaxPool2: $2\times2$, stride $=2$  
5. Conv3: $3\times3$, stride $=1$, padding $=1$, filters $=128$  
6. MaxPool3: $2\times2$, stride $=2$

### Step-by-step size changes
- **Input:** $128 \times 128 \times 3$
- **After Conv1:**
  $$
  \left\lfloor \frac{128 + 2(1) - 3}{1} \right\rfloor + 1 = 128
  $$
  $\rightarrow$ $128 \times 128 \times 32$

- **After MaxPool1:**
  $$
  \left\lfloor \frac{128 - 2}{2} \right\rfloor + 1 = 64
  $$
  $\rightarrow$ $64 \times 64 \times 32$

- **After Conv2:** spatial stays same (padding $=1$, stride $=1$)
  $\rightarrow$ $64 \times 64 \times 64$

- **After MaxPool2:**
  $\rightarrow$ $32 \times 32 \times 64$

- **After Conv3:**
  $\rightarrow$ $32 \times 32 \times 128$

- **After MaxPool3:**
  $\rightarrow$ $16 \times 16 \times 128$

Final feature map size $= 16 \times 16 \times 128$.
If flattened before FC: total features $= 16\times16\times128 = 32768$.

````python
# Code Answer 09: dimension calculator
import math

def out_size(n, f, p=0, s=1):
    return math.floor((n + 2*p - f) / s) + 1

h = w = 128
c = 3
print(f'Input: {h}x{w}x{c}')

# Conv1
h = out_size(h, f=3, p=1, s=1)
w = out_size(w, f=3, p=1, s=1)
c = 32
print(f'After Conv1: {h}x{w}x{c}')

# Pool1
h = out_size(h, f=2, p=0, s=2)
w = out_size(w, f=2, p=0, s=2)
print(f'After Pool1: {h}x{w}x{c}')

# Conv2
h = out_size(h, f=3, p=1, s=1)
w = out_size(w, f=3, p=1, s=1)
c = 64
print(f'After Conv2: {h}x{w}x{c}')

# Pool2
h = out_size(h, f=2, p=0, s=2)
w = out_size(w, f=2, p=0, s=2)
print(f'After Pool2: {h}x{w}x{c}')

# Conv3
h = out_size(h, f=3, p=1, s=1)
w = out_size(w, f=3, p=1, s=1)
c = 128
print(f'After Conv3: {h}x{w}x{c}')

# Pool3
h = out_size(h, f=2, p=0, s=2)
w = out_size(w, f=2, p=0, s=2)
print(f'After Pool3: {h}x{w}x{c}')

print('Flattened size:', h * w * c)
````

# Question 10: [ Marks 20 ]

## Design and implement a deep Convolutional Neural Network (CNN) for a multi-class image classification problem using PyTorch. The network should consist of four convolutional blocks, where each block includes two convolution layers followed by activation and normalization, and a pooling layer at the end. The number of filters should increase progressively across blocks. After the convolutional part, flatten the feature maps and connect them to fully connected layers with dropout for regularization, and produce final outputs corresponding to K classes. Also explain the role of each part of the network in detail.

# Write Answer 10:

A suitable deep CNN for multi-class classification has **four convolution blocks** and a **fully connected classifier** at the end.

### Architecture design
Let input be $(B, 3, H, W)$ and classes be $K$.

- **Block 1:** $\mathrm{Conv}(3\rightarrow32) \rightarrow \mathrm{BN} \rightarrow \mathrm{ReLU} \rightarrow \mathrm{Conv}(32\rightarrow32) \rightarrow \mathrm{BN} \rightarrow \mathrm{ReLU} \rightarrow \mathrm{MaxPool}(2)$
- **Block 2:** $\mathrm{Conv}(32\rightarrow64) \rightarrow \mathrm{BN} \rightarrow \mathrm{ReLU} \rightarrow \mathrm{Conv}(64\rightarrow64) \rightarrow \mathrm{BN} \rightarrow \mathrm{ReLU} \rightarrow \mathrm{MaxPool}(2)$
- **Block 3:** $\mathrm{Conv}(64\rightarrow128) \rightarrow \mathrm{BN} \rightarrow \mathrm{ReLU} \rightarrow \mathrm{Conv}(128\rightarrow128) \rightarrow \mathrm{BN} \rightarrow \mathrm{ReLU} \rightarrow \mathrm{MaxPool}(2)$
- **Block 4:** $\mathrm{Conv}(128\rightarrow256) \rightarrow \mathrm{BN} \rightarrow \mathrm{ReLU} \rightarrow \mathrm{Conv}(256\rightarrow256) \rightarrow \mathrm{BN} \rightarrow \mathrm{ReLU} \rightarrow \mathrm{MaxPool}(2)$
- **Classifier:** $\mathrm{Flatten} \rightarrow \mathrm{Linear} \rightarrow \mathrm{ReLU} \rightarrow \mathrm{Dropout} \rightarrow \mathrm{Linear} \rightarrow \mathrm{ReLU} \rightarrow \mathrm{Dropout} \rightarrow \mathrm{Linear}(128\rightarrow K)$

### Why each part is used
- **Convolution layers:** learn local spatial features (edges, textures, shapes).
- **Progressive filters ($32\to64\to128\to256$):** deeper blocks learn richer and more abstract patterns.
- **BatchNorm:** stabilizes activations, speeds up convergence.
- **ReLU:** adds nonlinearity and reduces vanishing-gradient effect.
- **MaxPool:** reduces spatial size and computation, keeps strong activations.
- **Flatten + FC layers:** combine extracted features for final class decision.
- **Dropout:** regularization to reduce overfitting.
- **Final $\mathrm{Linear}(128\rightarrow K)$:** produces logits for `CrossEntropyLoss`.

### Training flow (brief)
1. Forward pass to get logits.
2. Compute `CrossEntropyLoss(logits, labels)`.
3. `loss.backward()` for gradients.
4. `optimizer.step()` to update parameters.

This design is standard, deep enough for complex image patterns, and regularized with BatchNorm + Dropout.

````python
# Code Answer 10: Deep CNN in PyTorch (4 convolution blocks)
import torch
import torch.nn as nn

class MyCNN(nn.Module):
    def __init__(self, num_classes=10):
        super().__init__()

        self.features = nn.Sequential(
            # Block 1
            nn.Conv2d(3, 32, kernel_size=3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(inplace=True),
            nn.Conv2d(32, 32, kernel_size=3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2),

            # Block 2
            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True),
            nn.Conv2d(64, 64, kernel_size=3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2),

            # Block 3
            nn.Conv2d(64, 128, kernel_size=3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(inplace=True),
            nn.Conv2d(128, 128, kernel_size=3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2),

            # Block 4
            nn.Conv2d(128, 256, kernel_size=3, padding=1),
            nn.BatchNorm2d(256),
            nn.ReLU(inplace=True),
            nn.Conv2d(256, 256, kernel_size=3, padding=1),
            nn.BatchNorm2d(256),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(2),
        )

        # For input 128x128, after 4 pools => 8x8
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(256 * 8 * 8, 512),
            nn.ReLU(inplace=True),
            nn.Dropout(0.5),
            nn.Linear(512, 128),
            nn.ReLU(inplace=True),
            nn.Dropout(0.3),
            nn.Linear(128, num_classes)
        )

    def forward(self, x):
        x = self.features(x)
        x = self.classifier(x)
        return x

# Quick test
K = 5
model = MyCNN(num_classes=K)
x = torch.randn(4, 3, 128, 128)
logits = model(x)
print('Output shape:', logits.shape)  # (4, K)

# Example training step
labels = torch.tensor([0, 1, 2, 3])
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)

optimizer.zero_grad()
loss = criterion(logits, labels)
loss.backward()
optimizer.step()
print('Sample loss:', float(loss))
````
