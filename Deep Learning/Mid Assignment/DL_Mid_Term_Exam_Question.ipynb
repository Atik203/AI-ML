### Q3(c) Using Autograd (Actual Code)

```python
import torch

# x = 3 with gradient tracking
x = torch.tensor(3.0, requires_grad=True)

# Forward computation
y1 = x**2
y2 = 4*x
L = y1 + y2

# Backward computation
L.backward()

print("x =", x.item())
print("L =", L.item())
print("dL/dx =", x.grad.item())  # expected 10.0
```

- Autograd gives gradient directly after `backward()`.
- Here result is $\frac{dL}{dx}=10$.

# Question 04: [ Marks 15 ]

Create a scalar tensor x = 3.0 with gradient tracking enabled. <br>

Perform backpropagation. <br>
Print the gradient of x. <br>

## Write Answer 04:

```python
import torch

# 1) Scalar tensor with gradient tracking
x = torch.tensor(3.0, requires_grad=True)

# 2) Compute y = x^2 + 4x
y = x**2 + 4*x

# 3) Backpropagation
y.backward()

# 4) Print gradient of x
print("x.grad =", x.grad.item())  # Expected: 10.0
```

- Here $\frac{dy}{dx}=2x+4$.
- At $x=3$, gradient is $10$.

# Question 05: [ Marks 10 ]

Create two tensors:
w = 2.0 (requires_grad=True) <br>
b = 1.0 (requires_grad=True) <br>

Let input x = 3.0. <br>
Compute z = w*x + b. <br>
Use Sigmoid to get y_pred.
 <br>

Define Binary Cross Entropy manually (without built-in loss). <br>
Compute gradients using backward(). <br>
Print gradients of w and b. <br>

Write full code.

## Write Answer 05:

```python
import torch

# Parameters with gradients
w = torch.tensor(2.0, requires_grad=True)
b = torch.tensor(1.0, requires_grad=True)

# Input and target
x = torch.tensor(3.0)
y = torch.tensor(1.0)  # class label

# Linear part: z = w*x + b
z = w*x + b

# Sigmoid prediction
y_pred = torch.sigmoid(z)

# Manual BCE loss: -[y*log(y_pred) + (1-y)*log(1-y_pred)]
eps = 1e-7  # avoid log(0)
loss = -(y*torch.log(y_pred + eps) + (1-y)*torch.log(1 - y_pred + eps))

# Backward
loss.backward()

print("y_pred =", y_pred.item())
print("loss =", loss.item())
print("dw =", w.grad.item())
print("db =", b.grad.item())
```

- We manually wrote BCE formula.
- `backward()` gives gradients for both `w` and `b`.

```python
import torch

# x with gradient tracking
x = torch.tensor(2.0, requires_grad=True)

# Compute terms
y1 = x**2
y2 = 3*x
L = y1 + y2

# Backward
L.backward()

print("x.grad =", x.grad.item())  # Expected: 7.0
```

# Question 07: [ Marks 20 ]


Create tensor w = 2.0 (requires_grad=True). <br>

Compute two different losses sequentially and call backward() twice. <br>
Print w.grad. <br>

Do not clear gradients. <br>
Explain what you observe. <br>

Write full code.


## Write Answer 07:

```python
import torch

w = torch.tensor(2.0, requires_grad=True)

# First loss
L1 = w**2
L1.backward()
print("After first backward, w.grad =", w.grad.item())  # 2w = 4

# Second loss (without clearing gradients)
L2 = 3*w
L2.backward()
print("After second backward, w.grad =", w.grad.item())  # 4 + 3 = 7
```

- Gradient is accumulated in PyTorch by default.
- After first backward: grad = 4.
- After second backward without zeroing: grad = 7.
- So, gradients add up if we do not clear them (`w.grad.zero_()`).
