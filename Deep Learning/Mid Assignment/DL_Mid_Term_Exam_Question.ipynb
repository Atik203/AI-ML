### Q3(c) Using Autograd (Actual Code)

- Expected output: `dL/dx = 10.0`.
- In Autograd, setting `requires_grad=True` tells PyTorch to track operations on `x`.
- During forward pass, PyTorch builds a computation graph for $y_1=x^2$, $y_2=4x$, and $L=y_1+y_2$.
- When we call `L.backward()`, Autograd applies the chain rule automatically and accumulates the result in `x.grad`.
- Here, $$\frac{dL}{dx}=\frac{d(x^2+4x)}{dx}=2x+4$$ and at $x=3$, $$\frac{dL}{dx}=2(3)+4=10$$.

````python
import torch

x = torch.tensor(3.0, requires_grad=True)
y1 = x**2
y2 = 4*x
L = y1 + y2

L.backward()

print("x =", x.item())
print("L =", L.item())
print("dL/dx =", x.grad.item())  
````

# Question 05: [ Marks 10 ]

Create two tensors:
w = 2.0 (requires_grad=True) <br>
b = 1.0 (requires_grad=True) <br>

Let input x = 3.0. <br>
Compute z = w*x + b. <br>
Use Sigmoid to get y_pred.
 <br>

Define Binary Cross Entropy manually (without built-in loss). <br>
Compute gradients using backward(). <br>
Print gradients of w and b. <br>

Write full code.

## Write Answer 05:

````python
import torch

w = torch.tensor(2.0, requires_grad=True)
b = torch.tensor(1.0, requires_grad=True)

x = torch.tensor(3.0)
y = torch.tensor(1.0)

z = w*x + b

y_pred = torch.sigmoid(z)

# Manual BCE loss
eps = 1e-7
loss = -(y*torch.log(y_pred + eps) + (1-y)*torch.log(1 - y_pred + eps))

loss.backward()

print("y_pred =", y_pred.item())
print("loss =", loss.item())
print("dw =", w.grad.item())
print("db =", b.grad.item())
````

# Question 07: [ Marks 20 ]


Create tensor w = 2.0 (requires_grad=True). <br>

Compute two different losses sequentially and call backward() twice. <br>
Print w.grad. <br>

Do not clear gradients. <br>
Explain what you observe. <br>

Write full code.


## Write Answer 07:


````python
import torch

w = torch.tensor(2.0, requires_grad=True)

# First loss
L1 = w**2
L1.backward()
print("After first backward, w.grad =", w.grad.item())  

# Second loss (without clearing gradients)
L2 = 3*w
L2.backward()
print("After second backward, w.grad =", w.grad.item()) 
````

PyTorch adds each new gradient on top of whatever is already in `w.grad` — it never resets automatically. This is called gradient accumulation. <br>
After 1st `backward()` → `w.grad = 4` (dL1/dw = 2w = 4) <br>
After 2nd `backward()` → `w.grad = 4 + 3 = 7` (added, not replaced) <br>
To avoid this we have to use optimizer.zero_grad() before each backward pass in a training loop.
