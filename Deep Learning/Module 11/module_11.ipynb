```python

y = torch.tensor(1.0,requires_grad=True)
z = 2*y + 1
z.backward()
print(y.grad)
```

```python
def binary_cross_entropy(pred, target):
    epsilon = 1e-8
    pred = torch.clamp(pred, epsilon, 1 - epsilon)
    return - (target * torch.log(pred) + (1 - target) * torch.log(1 - pred))
```
